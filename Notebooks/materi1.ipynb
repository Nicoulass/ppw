---

## 1. Apa itu Web Mining?

Secara sederhana, **Web Mining** itu seperti menambang emas, tapi yang kita tambang adalah informasi berharga dari internet menggunakan teknik *data mining*. Informasi yang didapat ini nantinya bakal sangat berguna buat perusahaan dalam ngambil keputusan strategis.

> *"Web mining itu intinya pakai teknik data mining buat nemuin dan mengekstrak info dari layanan web secara otomatis."* (Etzioni, 1996).

**Kenapa data web susah diproses?**
- Datanya luar biasa guede (jutaan website).
- Formatnya macem-macem (HTML, XML, teks biasa).
- Isinya sangat dinamis dan berubah tiap waktu.
- Nggak ada topik yang spesifik, semuanya kecampur.

---

## 2. Pembagian (Taksonomi) Web Mining

Web mining biasanya dibagi jadi tiga kategori utama:

### A. Web Content Mining (Nambang Isi Web)
Fokusnya adalah ngekstrak informasi berguna dari konten yang kelihatan di halaman web, misalnya:
- **Teks (Text Mining)** -> baca artikel berita, komen netizen.
- **Gambar (Image Mining)** -> ngolah foto, ilustrasi produk.
- **Audio/Video** -> nganalisis lagu, rekaman, atau klip YouTube.

**Contoh Aplikasinya:**
- *Sentiment analysis* (nyari tahu komentar ini positif atau negatif).
- Bikin ringkasan dokumen otomatis biar cepet baca.
- Deteksi email spam.

### B. Web Usage Mining (Analisis Perilaku Pengguna)
Kalau yang ini fokusnya nganalisis data yang ditinggalkan pengguna pas lagi asyik browsing (*clickstream*). Contohnya kayak data log server, atau *cookie* browser.

**Contoh Aplikasinya:**
- **Product Recommendation:** Ini alasan kenapa habis kita buka Shopee/Tokped, tiba-tiba muncul iklan barang yang persis kita cari.
- **Personalized Search:** Hasil pencarian Google yang udah nyesuain sama history browsing kita masing-masing.

### C. Web Structure Mining (Nambang Struktur Web)
Nah, kalau yang ini fokusnya nyari pola dari *link* atau *hyperlink* antar website, sama struktur sosial di internet.

**Contoh Aplikasinya:**
- **Google PageRank:** Algoritma yang nentuin web mana yang pantes ditaruh di halaman pertama pencarian.
- **Nyari Influencer:** Buat nyari tahu siapa akun pusat (paling ngaruh) di komunitas Twitter/Instagram (Centrality).

---

## 3. Tahapan Web Mining

Secara umum, alurnya mirip kayak data mining biasa, cuma beda di fase pengumpulan datanya:

1. **Pengumpulan Data:** Biasanya pakai proses *crawling* otomatis atau narik data lewat Web API.
2. **Preprocessing (Beres-beres Data):** Ini tahapan yang paling menguras tenaga dan waktu (bisa 70-80% dari total project). Teks dibersihkan, diubah, disaring, dan dipotong-potong biar komputer bisa baca dan olah.
3. **Data Mining & Evaluasi:** Ngejalanin metode *machine learning* dan mengevaluasi hasil pola dari data yang udah bersih.


## Praktik Tugas 1: Scraping Berita Detik

Di tugas ini, aku bakal nyoba ngambil (crawling/scraping) secara otomatis **200 berita** dari detik.com.
Rincian target pengambilannya:
- 100 berita dari kanal **detikSport**
- 100 berita dari kanal **detikFinance**

Nantinya semua hasil tarikan teks ini bakal aku kumpulin dan susun ke dalam format tabel (Pandas DataFrame) dengan susunan kolom: **ID**, **Isi Berita**, dan **Label**.

### Langkah 1: Siapin Library (Trafilatura)
Di sini aku nggak pakai *library scraping* yang rumit, tapi cukup pakai **Trafilatura**. Ini *library* ampuh yang emang didesain khusus buat otomatis ngambil teks utama dari halaman website berita.

```python
from trafilatura import fetch_url, extract
from trafilatura.feeds import find_feed_urls
import pandas as pd
import os
```

### Langkah 2: Mencari Otomatis Link Artikel
Karena sitemap detik.com lumayan kompleks, trik jitunya adalah nggunain link RSS Feed resmi mereka lewat fitur `find_feed_urls` milik Trafilatura. Dengan begini, aku langsung dapet 100 *link* berita paling hangat.

```python
print("Mencari link berita sport...")
sport_links = find_feed_urls('https://sport.detik.com/rss')[:100]

print("Mencari link berita finance...")
finance_links = find_feed_urls('https://finance.detik.com/rss')[:100]

data_berita = []
```

### Langkah 3: Eksekusi Looping 100 Berita SPORT (ID 1-100)
Sekarang waktunya kita *looping* (muter) satu-persatu link kategori Sport. Tiap kali dapet halamannya, teksnya diekstrak pakai `extract`, lalu dimasukin ke dalam *list* keranjang sementara bernama `data_berita`.

```python
for idx, link in enumerate(sport_links, start=1):
    downloaded = fetch_url(link)
    if downloaded:
        isi_berita = extract(downloaded)
        data_berita.append({
            "ID": idx,
            "Isi Berita": isi_berita,
            "Label": "sport"
        })
        print(f"[{idx}/100] Berhasil ambil: {link}")
```

### Langkah 4: Eksekusi Looping 100 Berita FINANCE (ID 101-200)
Sama persis kayak langkah di atas, tapi kali ini *looping* buat kategori Finance. Jangan lupa `start` (ID mulainya) diset ke angka 101 biar urutannya rapi lanjut dari ID 100.

```python
for idx, link in enumerate(finance_links, start=101):
    downloaded = fetch_url(link)
    if downloaded:
        isi_berita = extract(downloaded)
        
        data_berita.append({
            "ID": idx,
            "Isi Berita": isi_berita,
            "Label": "finance"
        })
        print(f"[{idx}/200] Berhasil ambil: {link}")
print(f"Selesai! Total berita terkumpul: {len(data_berita)}")
```

### Langkah 5: Tampilkan di Tabel (Pandas) & Simpan ke CSV
Semua datanya udah kumpul di *list*. Biar enak dilihat dan siap dipakai buat tugas selanjutnya, aku masukin datanya ke tabel `pandas` (DataFrame).

**Penting:** Setelah itu, tabelnya aku simpan (*export*) jadi format `.csv` ke dalam folder `output/`. Supaya nanti di Tugas 2 (materi2.ipynb) tinggal baca (*import*) dari file CSV ini aja!

```python
import os

df = pd.DataFrame(data_berita)

os.makedirs('output', exist_ok=True)

df.to_csv('output/crawling_detik.csv', index=False)

display(df)
```

### Langkah 6 : Menyimpan Hasil Ekstraksi ke File CSV
Menyimpan tabel berita yang telah di-scraping ke dalam folder `Output/berita.csv` agar dapat digunakan pada materi selanjutnya.

```python
os.makedirs("Output", exist_ok=True)

# Menyimpan DataFrame ke file CSV
df.to_csv("Output/berita.csv", index=False, encoding="utf-8")
print(f"Folder 'Output' siap.")
print(f"Data berhasil disimpan ke 'Output/berita.csv' (Total: {len(df)} baris)")
```

## Hasil Detail: Analisis Hasil Scraping

Bagian ini membaca kembali file CSV yang **sudah** dihasilkan di atas (tidak crawling ulang),
sehingga angka-angkanya bisa ditampilkan kapan saja tanpa takut data berubah.

- **Hasil A** - Ringkasan hasil scraping: jumlah berita per kelas, panjang teks, teks kosong
- **Hasil B** - Kualitas ekstraksi: panjang artikel & jumlah kata per kelas, artikel terpendek/terpanjang
- **Hasil C** - Konsistensi antar file CSV yang dihasilkan (ada 3 file dengan jumlah baris berbeda)

```python
from pathlib import Path
import pandas as pd

FILES = {
    "crawling_detik.csv (Langkah 5)": Path("Output/crawling_detik.csv"),
    "Output/berita.csv (Langkah 6)": Path("Output/berita.csv"),
    "../output/berita.csv (dipakai aplikasi)": Path("../output/berita.csv"),
}

print("=" * 72)
print("HASIL A : Ringkasan hasil scraping")
print("=" * 72)

frames = {}
for nama, path in FILES.items():
    if not path.exists():
        print(f"\n{nama}\n  -> file tidak ditemukan")
        continue
    d = pd.read_csv(path)
    frames[nama] = d
    kosong = int(d["Isi Berita"].isna().sum())
    s = d["Isi Berita"].dropna().astype(str).str.len()
    per_label = {k: int(c) for k, c in d['Label'].value_counts().items()}
    print(f"\n{nama}")
    print(f"  jumlah baris  : {len(d)}")
    print(f"  kolom         : {list(d.columns)}")
    print(f"  per label     : {per_label}")
    print(f"  teks kosong   : {kosong}")
    print(f"  panjang teks  : min={s.min()}, rata-rata={s.mean():.0f}, maks={s.max()}, total={s.sum():,} karakter")
```

```python
print("=" * 72)
print("HASIL B : Kualitas ekstraksi per kelas")
print("=" * 72)

d = frames["crawling_detik.csv (Langkah 5)"].copy()
d["karakter"] = d["Isi Berita"].astype(str).str.len()
d["kata"] = d["Isi Berita"].astype(str).str.split().str.len()

print("\nStatistik jumlah KARAKTER per dokumen:")
print(d.groupby("Label")["karakter"].agg(["count", "min", "median", "mean", "max"]).round(0).to_string())

print("\nStatistik jumlah KATA per dokumen:")
print(d.groupby("Label")["kata"].agg(["count", "min", "median", "mean", "max"]).round(0).to_string())

print("\nDistribusi panjang artikel (jumlah dokumen):")
bins = [0, 500, 1000, 2000, 3000, 5000, 10_000]
d["bucket"] = pd.cut(d["karakter"], bins=bins)
print(pd.crosstab(d["bucket"], d["Label"]).to_string())

print("\n3 artikel TERPENDEK:")
print(d.nsmallest(3, "karakter")[["ID", "Label", "karakter", "kata"]].to_string(index=False))
print("\n3 artikel TERPANJANG:")
print(d.nlargest(3, "karakter")[["ID", "Label", "karakter", "kata"]].to_string(index=False))

print(f"\nTotal kata seluruh korpus : {d['kata'].sum():,} kata")
print(f"Rata-rata per dokumen     : {d['kata'].mean():.0f} kata")
```

```python
print("=" * 72)
print("HASIL C : Konsistensi antar file CSV")
print("=" * 72)

rows = []
for nama, path in FILES.items():
    if path.exists():
        dd = pd.read_csv(path)
        vc = dd["Label"].value_counts()
        rows.append({
            "File": str(path),
            "Baris": len(dd),
            "sport": int(vc.get("sport", 0)),
            "finance": int(vc.get("finance", 0)),
            "Teks kosong": int(dd["Isi Berita"].isna().sum()),
        })

print(pd.DataFrame(rows).to_string(index=False))
print("""
Catatan:
- Output/crawling_detik.csv : 200 baris, hasil utama Langkah 5 (dipakai materi 2 & 5).
- Output/berita.csv         : 199 baris (finance tinggal 99) - selisih 1 baris dari
                              crawling_detik.csv, kemungkinan karena eksekusi parsial.
- ../output/berita.csv       : 200 baris, salinan yang dibaca train_model.py untuk
                              melatih model aplikasi yang di-deploy.
Pertanyaan dosen "pakai data yang mana?" -> jawab: 200 baris, 100 sport + 100 finance,
dan aplikasi memakai ../output/berita.csv.
""")
```
