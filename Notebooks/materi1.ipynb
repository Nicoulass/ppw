## Apa itu Web Mining?

**Web Mining** adalah penerapan teknik data mining untuk mengekstrak dan menemukan informasi yang berguna dari data web. Web mining mencakup tiga kategori utama:

1. **Web Content Mining** — Menganalisis konten halaman web (teks, gambar, dll)
2. **Web Structure Mining** — Menganalisis struktur hyperlink antar halaman web
3. **Web Usage Mining** — Menganalisis pola akses pengguna dari log server
