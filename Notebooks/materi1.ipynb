## 1. Definisi Web Mining

**Penambangan web (Web Mining)** adalah proses menemukan dan mengekstrak informasi dari internet menggunakan berbagai teknik penambangan data. Informasi ini dapat digunakan oleh perusahaan untuk pengambilan keputusan yang efektif.

> *"Penambangan web adalah penggunaan teknik penambangan data untuk secara otomatis menemukan dan mengekstrak informasi dari layanan web"* (Etzioni, 1996; CACM 39).

> *"Penambangan web bertujuan untuk menemukan pola-pola berguna atau pengetahuan dari struktur hiperlink web, isi halaman web, serta perilaku pengguna."* (Bing Liu, 2007, Web Data Mining).

### Tantangan Pemrosesan Data Web
- Web adalah pangkalan data yang sangat besar.
- Datanya dalam format HTML, XML, text.
- Sangat kompleks dan dinamis.
- Bukan domain yang spesifik.

---

## 2. Taksonomi Web Mining

Web mining dibagi menjadi tiga subbidang utama:

### A. Penambangan Isi Web (Web Content Mining)
Proses mengekstrak informasi berguna dari dokumen-dokumen web. Fokus pada konten halaman web seperti:
- **Teks (Text Mining)** -> Artikel, deskripsi, komentar
- **Gambar (Image Mining)** -> Foto, ilustrasi
- **Audio (Audio Mining)** -> Rekaman Suara, Musik
- **Video (Video Mining)** -> Streaming clip
- **Data terstruktur** -> Tabel, daftar

**Aplikasi Text Mining:**
- Ekstraksi Informasi
- Topic modelling
- Peringkasan dokumen
- Klasifikasi dokumen (sentiment analysis, opinion mining, kategorisasi berita, deteksi spam)
- Pengelompokan konten (clustering hasil pencarian)

### B. Web Usage Mining (Analisis Clickstream)
Ekstraksi informasi dari data yang dihasilkan melalui kunjungan halaman web dan transaksi. Data biasanya disimpan dalam log akses server, log referer, log agen, dan cookie.

**Aplikasi:**
- **Product Recommendation:** Sistem yang memprediksi dan menyarankan produk yang diminati pengguna berdasarkan perilaku.
- **Personalized Search:** Pencarian yang disesuaikan dengan preferensi dan konteks pengguna individu.

### C. Penambangan Struktur Web (Web Structure Mining)
Penemuan dan interpretasi pola dalam struktur tautan (hyperlink) di internet dan hubungan sosial antar pelaku di web.

**Aplikasi:**
- **Google Search:** Menggunakan algoritma *PageRank* untuk menentukan urutan hasil pencarian berdasarkan otoritas halaman.
- **Twitter/Instagram:** Menampilkan rekomendasi akun influencer (Centrality).
- **Deteksi Komunitas:** Mencari pemimpin komunitas atau penyebar berita palsu, serta sistem rekomendasi berbasis komunitas.

---

## 3. Proses Web Mining

Secara umum mirip dengan proses data mining, namun berbeda pada tahapan pengumpulan data:

1. **Gathering and Exploration (Pengumpulan Data)**
   - Crawling dokumen/data
   - Pengambilan data menggunakan Web API
   - Memahami dan memvisualisasikan data
2. **Preprocessing and Transformation**
   - Mentransformasi data ke representasi yang sesuai (reduksi dimensi, seleksi fitur, tokenisasi teks, embedding).
   - Memakan waktu 70-80% dari total proyek data mining.
3. **Actual Data Mining**
   - Menggunakan metode data mining.
   - Mengevaluasi model/pola yang dihasilkan.
   - Melakukan iterasi (eksperimen hyperparameter, perbaikan praproses data).

### Langkah 1 : Import Library Trafilatura

```python
from trafilatura import fetch_url, extract
from trafilatura.feeds import find_feed_urls
import pandas as pd
```

### Langkah 2 : Mencari otomatis link artikel menggunakan fitur bawaan trafilatura

```python
print("Mencari link berita sport...")
sport_links = find_feed_urls('https://sport.detik.com/rss')[:100]

print("Mencari link berita finance...")
finance_links = find_feed_urls('https://finance.detik.com/rss')[:100]

data_berita = []
```

### Langkah 3 : Looping untuk 100 berita SPORT (ID 1-100)

```python
for idx, link in enumerate(sport_links, start=1):
    downloaded = fetch_url(link)
    if downloaded:
        isi_berita = extract(downloaded)
        data_berita.append({
            "ID": idx,
            "Isi Berita": isi_berita,
            "Label": "sport"
        })
        print(f"[{idx}/100] Berhasil ambil: {link}")
```

### Langkah 4 : Looping untuk 100 berita FINANCE (ID 101-200)

```python
for idx, link in enumerate(finance_links, start=101):
    downloaded = fetch_url(link)
    if downloaded:
        isi_berita = extract(downloaded)
        
        data_berita.append({
            "ID": idx,
            "Isi Berita": isi_berita,
            "Label": "finance"
        })
        print(f"[{idx}/200] Berhasil ambil: {link}")
print(f"Selesai! Total berita terkumpul: {len(data_berita)}")
```

### Langkah 5 : Menampilkan Data dalam Bentuk Tabel

```python
df = pd.DataFrame(data_berita)

display(df)
```
